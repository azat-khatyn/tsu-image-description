```python
import matplotlib.pyplot as plt

epochs = [0, 1]

train_loss = [
    0.580497930047801,
    0.008251301109537118
]

val_loss = [
    0.011297855482278567,
    0.009354914460585997
]

plt.figure()
plt.plot(epochs, train_loss, marker='o', label="Train Loss")
plt.plot(epochs, val_loss, marker='o', label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Dynamics of BLIP Fine-tuning")
plt.legend()
plt.grid()

plt.savefig("loss_curve.png")
plt.show()
```
