## Ячейка 1. Все 14 семантических тестов

Для каждого изображения: эталон + сгенерированное архивное описание + ключевые метрики.

```python
import json
from pathlib import Path
from IPython.display import display, Markdown, Image

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
SEMTEST = ROOT / 'data/eval/results/metrics_PROPOSED_semtest.json'

with open(SEMTEST) as f:
    d = json.load(f)
items = d['per_item']

agg = d['summary']['metrics']
ret = d['summary'].get('retrieval', {}).get('t2i', {})
display(Markdown(
    f"**Aggregate (n={len(items)}):**  "
    f"CLIPScore_EN = {agg.get('CLIPScore_EN_caption_en',0):.4f}  · "
    f"CLIPScore_RU(caption) = {agg.get('CLIPScore_RU_caption_ru',0):.4f}  · "
    f"CLIPScore_RU(archive) = {agg.get('CLIPScore_RU_archive_ru',0):.4f}  · "
    f"R@1 = {ret.get('R@1','—'):.2f}  · "
    f"R@5 = {ret.get('R@5','—'):.2f}"
))

for i, it in enumerate(items, 1):
    img_path = ROOT / it['image_path']
    fn = Path(it['image_path']).name
    score_arch = it['scores'].get('CLIPScore_RU_archive_ru', 0)
    score_cap = it['scores'].get('CLIPScore_RU_caption_ru', 0)
    display(Markdown(f"### {i}. {fn}  ·  *[{it.get('source','?')}]*"))
    if img_path.exists():
        display(Image(filename=str(img_path), width=280))
    display(Markdown(
        f"- **REF:** {it.get('reference_ru') or '—'}\n"
        f"- **caption_en:** {it.get('caption_en') or '—'}\n"
        f"- **caption_ru:** {it.get('caption_ru') or '—'}\n"
        f"- **archive_ru:** {it.get('archive_ru') or '—'}\n"
        f"- *CLIPScore_RU(archive)* = {score_arch:.4f}  ·  *CLIPScore_RU(caption)* = {score_cap:.4f}"
    ))
    display(Markdown('---'))
```

## Ячейка 2. Top-N по CLIPScore_RU(archive) на НЭБ-коллекции с разметкой

Из 224 размеченных открыток коллекции №529 РНБ — топ-N по релевантности архивного описания изображению (по CLIPScore_RU).

```python
import json
from pathlib import Path
from IPython.display import display, Markdown, Image

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
NEB = ROOT / 'data/eval/results/metrics_E12_neb_n224.json'
TOP_N = 100  # измени при необходимости

with open(NEB) as f:
    d = json.load(f)
items = d['per_item']

agg = d['summary']['metrics']
ret = d['summary'].get('retrieval', {}).get('t2i', {})
display(Markdown(
    f"**Aggregate (n={len(items)}):**  "
    f"CLIPScore_EN = {agg.get('CLIPScore_EN_caption_en',0):.4f}  · "
    f"CLIPScore_RU(caption) = {agg.get('CLIPScore_RU_caption_ru',0):.4f}  · "
    f"CLIPScore_RU(archive) = {agg.get('CLIPScore_RU_archive_ru',0):.4f}  · "
    f"R@1 = {ret.get('R@1','—'):.3f}  · "
    f"R@5 = {ret.get('R@5','—'):.3f}"
))

ranked = sorted(items, key=lambda it: it['scores'].get('CLIPScore_RU_archive_ru', 0), reverse=True)
top = ranked[:TOP_N]

display(Markdown(f"### Top-{TOP_N} по CLIPScore_RU(archive)"))
for i, it in enumerate(top, 1):
    img_path = ROOT / it['image_path']
    fn = Path(it['image_path']).name
    score_arch = it['scores'].get('CLIPScore_RU_archive_ru', 0)
    display(Markdown(f"### {i}. {fn}  ·  CLIPScore = **{score_arch:.4f}**"))
    if img_path.exists():
        display(Image(filename=str(img_path), width=280))
    display(Markdown(
        f"- **REF (curator, RUSMARC 327):** {it.get('reference_ru') or '—'}\n"
        f"- **archive_ru (пайплайн):** {it.get('archive_ru') or '—'}"
    ))
    display(Markdown('---'))
```
